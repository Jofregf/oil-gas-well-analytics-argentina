## 6. Missing Values Assessment

### Business relevance

Missing values may reduce data quality, affect statistical analyses,
introduce bias, and limit the variables available for predictive modeling.
Understanding missing data is essential before designing the analytical
workflow.

### Objective

Assess the extent and distribution of missing values in both datasets to
identify variables that may require cleaning, imputation, or exclusion
from subsequent analyses.

### Hypothesis

Core production variables are expected to contain few or no missing values,
whereas descriptive and optional attributes may present higher proportions
of missing data.

```python
def missing_values_summary(df):
    """
    Generate a summary of missing values for each column.

    Parameters
    ----------
    df : pandas.DataFrame
        Input DataFrame.

    Returns
    -------
    pandas.DataFrame
        Summary containing the number and percentage of missing values
        for each column.
    """

    missing = df.isna().sum()
    summary_df = (
        pd.DataFrame({
            "Missing Values": missing,
            "Missing %": ((missing / len(df)) * 100).round(2)
        })
        .rename_axis("Column")
        .sort_values("Missing %", ascending=False)
    )

    return summary_df
```

```python
production_missing = missing_values_summary(production_sample)
production_missing[production_missing["Missing Values"] > 0]
```

```python
fracture_missing = missing_values_summary(fracture_df)
fracture_missing[fracture_missing["Missing Values"] > 0]
```

## Findings

- The production dataset is highly complete, with only two variables containing missing values.
- `observaciones` and `vida_util` each present approximately 99.9% missing values, suggesting they may be optional or conditionally reported attributes.
- The fracture dataset also exhibits high completeness.
- `tipo_reservorio` is nearly complete (0.35% missing), whereas `subtipo_reservorio` has a substantially higher proportion of missing values (30.30%).
- The relationship between `tipo_reservorio` and `subtipo_reservorio` should be investigated in subsequent analyses.

### Business relevance

Reservoir classification is an important geological attribute that may

### Objective

Investigate the relationship between `tipo_reservorio` and
`subtipo_reservorio` to determine the possible causes of missing values.

### Hypothesis

Missing values in `subtipo_reservorio` are expected to be associated with
specific reservoir types rather than being randomly distributed across
all records.

```python
fracture_df["tipo_reservorio"].value_counts()
```

```python
pd.crosstab(
    fracture_df["tipo_reservorio"],
    fracture_df["subtipo_reservorio"]
)
```

```python
fracture_df[
    fracture_df["subtipo_reservorio"].isna()
]["tipo_reservorio"].value_counts(dropna=False)
```

### Findings

- Further investigation showed that all missing `subtipo_reservorio` values correspond either to:

  - records where `tipo_reservorio` is also missing (7 records).
- All `NO CONVENCIONAL` wells have a valid `subtipo_reservorio` classification (`SHALE` or `TIGHT`).
- Therefore, missing values in `subtipo_reservorio` are primarily explained by business rules rather than data quality issues.

## Findings

- The fracture dataset contains 4,562 unique wells, while the production sample contains 1,954 unique wells.
- A total of 1,570 unique wells are present in both datasets, confirming that `idpozo` can be used as the primary key to integrate fracture and production information.
- Within the current production sample, approximately 34.4% of fracture wells have matching production records.
- Conversely, about 80.3% of the wells in the production sample have corresponding fracture information.
- These percentages should be interpreted with caution because the production dataset analyzed is only a sample of the complete dataset. The overlap will be reassessed once the full production dataset is loaded.

## Conclusion
The audit confirms that both datasets share a common identifier (`idpozo`) and can be integrated for subsequent analyses. However, because the current production data represents only a sample, the completeness of the relationship between production and fracture records should be re-evaluated using the full production dataset before drawing definitive conclusions.
