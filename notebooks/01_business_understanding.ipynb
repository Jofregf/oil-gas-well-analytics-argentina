# 1. Project Overview

## Objective

Provide the business context of the project and explain why analyzing well production and hydraulic fracturing data can generate valuable operational insights.

## Overview

The Argentine oil and gas industry has experienced significant growth in unconventional hydrocarbon production, particularly in shale formations such as Vaca Muerta. Hydraulic fracturing plays a key role in the development of these resources, generating large volumes of operational and production data.

This project aims to integrate official production and hydraulic fracturing datasets to investigate how completion characteristics relate to well performance over time. The analysis combines SQL, Python, and Power BI to transform raw operational data into actionable business insights.

The project was designed to simulate a real-world analytics workflow using publicly available government data. Beyond exploring production trends, it provides an opportunity to apply data auditing, SQL analytics, Python-based data analysis, and business intelligence techniques to an industrial domain with growing strategic importance in Argentina.

# 2. Business Problem

## Objective

Define the business problem that motivates the project and explain why understanding the relationship between hydraulic fracturing operations and well production is valuable for decision-making.

## Business Context

Hydraulic fracturing represents one of the largest operational investments in unconventional oil and gas production. However, production performance varies significantly between wells due to geological, operational, and completion-related factors.

Understanding how fracture characteristics, well location, reservoir type, and operator influence production can help identify patterns associated with better-performing wells and support more informed operational and investment decisions.

## Business Problem

Oil and gas operators need to understand which operational and geological factors are associated with higher production performance. Although large volumes of production and hydraulic fracturing data are available, these datasets are often analyzed separately, limiting the ability to identify relationships between well completion practices and production outcomes.

This project addresses this challenge by integrating official production and hydraulic fracturing datasets to support data-driven decision-making and generate actionable insights into well performance.

#### Production Performance

- BQ-01 How does well production evolve over time?

#### Hydraulic Fracturing

- BQ-07 Is hydraulic fracturing associated with higher production performance?
- BQ-08 Which fracture characteristics are associated with higher production?
- BQ-09 Are there fractured wells without recorded production?

#### Geological Analysis

- BQ-10 Do different geological formations exhibit different production behaviors?
- BQ-11 Does reservoir type influence production performance?

- BQ-13 Which operators achieve the highest production?
